# Imports

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

# Get Data

```python
rewards_file_human = "episode_rewards_human.csv"
rewards_file_ppo = "episode_rewards_PPO.csv"
rewards_file_random = "episode_rewards_random.csv"
# rewards_file_sac = ""
```

```python
rewards_df_human = pd.read_csv(rewards_file_human)
rewards_df_ppo = pd.read_csv(rewards_file_ppo)
rewards_df_random = pd.read_csv(rewards_file_random)
# rewards_df_sac = pd.read_csv(rewards_file_sac)
```

```python
rewards_df_list = [
    rewards_df_human, 
    rewards_df_ppo, 
    rewards_df_random, 
    # rewards_df_sac
]

for index, reward_df in enumerate(rewards_df_list):
    if len(reward_df) > 0:
        rewards_df_list[index] = reward_df[:25]
        print(reward_df.head())
```

```python
for reward_df in rewards_df_list:
    print(len(reward_df))
```

# Visualize Data

```python
n = len(rewards_df_list)
titles = [
    "Human Agent", 
    "PPO Agent",
    "Random Agent",
]

fig, axes = plt.subplots(1, n, figsize=(6*n, 4), sharey=True)

# If there's only one DF, axes won't be a list
if n == 1:
    axes = [axes]

i = 0
for ax, reward_df in zip(axes, rewards_df_list):
    ax.plot(reward_df["episode"], reward_df["reward"])
    ax.set_title(f"Reward vs Episode: {titles[i]}")
    ax.set_xlabel("Episode")
    ax.set_ylabel("Reward")
    i += 1

plt.tight_layout()
plt.show()
```

```python
print("25 episode average rewards:")
for index, reward_df in enumerate(rewards_df_list):
    print(f"- {titles[index]}:", np.mean(reward_df["reward"]))
```
